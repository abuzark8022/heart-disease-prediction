## Code – Heart Disease Prediction System

### Step 1: Import Libraries

````python
"""
Purpose of this section:
-------------------------
This script imports the required libraries for building, benchmarking,
and evaluating an end-to-end machine learning system. Tools include
data manipulation (pandas, numpy), data preprocessing (LabelEncoder),
model training (SVC, Logistic Regression, Decision Tree), comprehensive
evaluation metrics, PrettyTable for clean formatting, and Seaborn/Matplotlib
for statistical data visualizations.
"""

# Numerical computing library
import numpy as np

# Data manipulation and analysis library
import pandas as pd

# Model serialization library
import pickle

# Dataset splitting utility
from sklearn.model_selection import train_test_split

# Categorical label encoder
from sklearn.preprocessing import LabelEncoder

# Primary Support Vector Machine Classifier
from sklearn import svm

# Benchmark classifiers for comparative evaluation
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

# Model evaluation metrics
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Clean tabular output formatting
from prettytable import PrettyTable

# Plotting and visualization libraries
import matplotlib.pyplot as plt
import seaborn as sns

print("All required libraries imported successfully.")
````

### Step 2: Load Sample Data

````python
"""
Purpose of this section:
-------------------------
This part loads the clean Heart Disease dataset from 'sample-data.csv'
into a pandas DataFrame and displays the instances.
"""

# Load the dataset
sample_data = pd.read_csv("sample-data.csv")

print("\nSample Data (First 15 Instances):")
print("==================================\n")
print(sample_data.head(15).to_string())
````

### Step 3: Understand and Pre-process Sample Data

#### Step 3.1: Understand Sample Data

````python
"""
Purpose of this section:
-------------------------
Display column attributes and total instance count in the dataset.
"""

print("\nAttributes in Sample Data:")
print("==========================\n")
print(sample_data.columns)

print("\nNumber of Instances in Sample Data:", sample_data["AgeGroup"].count())
print("========================================\n")
````

#### Step 3.2: Pre-process Sample Data
- Data has been verified and contains **zero missing values** across all selected attributes.
- Categorical attributes are clean and ready for label encoding.

````python
# Verify missing values
print("Missing values per column:\n", sample_data.isnull().sum())
print("\nTarget Class Distribution:\n", sample_data['HeartDisease'].value_counts())
````

#### Step 3.3: Visualize Heart Disease Distribution by Clinical Risk Factors (NEW)

````python
"""
Purpose of this section:
-------------------------
Data visualization: Examine how Heart Disease correlates with Chest Pain Type,
Sex, and Age Group prior to training the model.
"""

plt.figure(figsize=(14, 4))

plt.subplot(1, 3, 1)
sns.countplot(data=sample_data, x="ChestPain", hue="HeartDisease", palette="Set1")
plt.title("Heart Disease by Chest Pain Type")
plt.xticks(rotation=20)

plt.subplot(1, 3, 2)
sns.countplot(data=sample_data, x="Sex", hue="HeartDisease", palette="Set2")
plt.title("Heart Disease by Biological Sex")

plt.subplot(1, 3, 3)
sns.countplot(data=sample_data, x="AgeGroup", hue="HeartDisease", palette="Set3")
plt.title("Heart Disease by Age Group")

plt.tight_layout()
plt.show()
````

### Step 4: Feature Extraction and Selection (4 Cardinal Attributes)
- **Assignment Constraint Adherence:** Exactly **4 input attributes** are selected from the 14 attributes in the UCI Heart Disease dataset:
  1. `AgeGroup` (Derived from attribute #3: age)
  2. `Sex` (Attribute #4: biological sex)
  3. `ChestPain` (Attribute #9: cp, chest pain type)
  4. `ExerciseAngina` (Attribute #38: exang, exercise-induced angina)
- **Target:** `HeartDisease` (Attribute #58: num, angiographic diagnosis)
- **Clinical Rationale:** These four non-invasive markers represent the fundamental triage criteria checked by cardiologists before ordering invasive angiograms.

### Step 5: Label Encoding the Sample Data (Input and Output is converted in Numeric Representation)

#### Step 5.1: Train the Label Encoder

````python
"""
Purpose of this section:
-------------------------
Initialize and fit LabelEncoders on pre-defined, fixed category lists.
This ensures deterministic numeric mapping identical to the Titanic model pattern.
"""

# Define category lists
age_group = pd.DataFrame({"AgeGroup": ["Young", "MiddleAged", "Senior"]})
sex = pd.DataFrame({"Sex": ["Male", "Female"]})
chest_pain = pd.DataFrame({"ChestPain": ["TypicalAngina", "AtypicalAngina", "NonAnginal", "Asymptomatic"]})
exercise_angina = pd.DataFrame({"ExerciseAngina": ["Yes", "No"]})
heart_disease = pd.DataFrame({"HeartDisease": ["Yes", "No"]})

# Initialize LabelEncoders
age_group_label_encoder = LabelEncoder()
sex_label_encoder = LabelEncoder()
chest_pain_label_encoder = LabelEncoder()
exercise_angina_label_encoder = LabelEncoder()
heart_disease_label_encoder = LabelEncoder()

# Train (fit) each encoder on its respective categories
age_group_label_encoder.fit(np.ravel(age_group))
sex_label_encoder.fit(np.ravel(sex))
chest_pain_label_encoder.fit(np.ravel(chest_pain))
exercise_angina_label_encoder.fit(np.ravel(exercise_angina))
heart_disease_label_encoder.fit(np.ravel(heart_disease))

print("All 5 LabelEncoders trained successfully.")
````

#### Step 5.2: Label Encode the Output

````python
"""
Purpose of this section:
-------------------------
Apply Label Encoding to the target variable ("HeartDisease"): "No" -> 0, "Yes" -> 1.
Save the intermediate dataset to 'sample-data-encoded-output.csv'.
"""

sample_data_encoded_output = sample_data.copy()
original_sample_data = sample_data.copy()

print("\nHeartDisease Attribute After Label Encoding:")
print("============================================\n")
sample_data["encoded_heart_disease"] = heart_disease_label_encoder.transform(sample_data["HeartDisease"])
print(sample_data[["HeartDisease", "encoded_heart_disease"]].head(10))

sample_data_encoded_output[["AgeGroup", "Sex", "ChestPain", "ExerciseAngina", "HeartDisease"]] = \
    sample_data[["AgeGroup", "Sex", "ChestPain", "ExerciseAngina", "encoded_heart_disease"]]

sample_data_encoded_output = sample_data_encoded_output[["AgeGroup", "Sex", "ChestPain", "ExerciseAngina", "HeartDisease"]]
sample_data_encoded_output.to_csv("sample-data-encoded-output.csv", index=False, header=True)
print("\nSaved: sample-data-encoded-output.csv")
````

#### Step 5.3: Label Encode the Input

````python
"""
Purpose of this section:
-------------------------
Apply Label Encoding to the 4 input features.
Save the completely encoded dataset to 'sample-data-encoded.csv'.
"""

sample_data_encoded = sample_data_encoded_output.copy()

sample_data_encoded_output["encoded_age_group"] = age_group_label_encoder.transform(sample_data_encoded_output["AgeGroup"])
sample_data_encoded_output["encoded_sex"] = sex_label_encoder.transform(sample_data_encoded_output["Sex"])
sample_data_encoded_output["encoded_chest_pain"] = chest_pain_label_encoder.transform(sample_data_encoded_output["ChestPain"])
sample_data_encoded_output["encoded_exercise_angina"] = exercise_angina_label_encoder.transform(sample_data_encoded_output["ExerciseAngina"])

sample_data_encoded[["AgeGroup", "Sex", "ChestPain", "ExerciseAngina", "HeartDisease"]] = \
    sample_data_encoded_output[["encoded_age_group", "encoded_sex", "encoded_chest_pain", "encoded_exercise_angina", "HeartDisease"]]

sample_data_encoded = sample_data_encoded[["AgeGroup", "Sex", "ChestPain", "ExerciseAngina", "HeartDisease"]]

print("\nSample Data after Label Encoding (First 10 Rows):")
print("==================================================\n")
print(sample_data_encoded.head(10))

sample_data_encoded.to_csv("sample-data-encoded.csv", index=False, header=True)
print("\nSaved: sample-data-encoded.csv")
````

### Step 6: Execute the Training Phase

#### Step 6.1: Splitting Sample Data into Training Data and Testing Data

````python
"""
Purpose of this section:
-------------------------
Divide the dataset into Training Data (80%) and Testing Data (20%).
random_state=0 and shuffle=False ensure exact reproducibility.
"""

training_data_encoded, testing_data_encoded = train_test_split(
    sample_data_encoded,
    test_size=0.2,
    random_state=0,
    shuffle=False
)

training_data_encoded.to_csv("training-data-encoded.csv", index=False, header=True)
testing_data_encoded.to_csv("testing-data-encoded.csv", index=False, header=True)

print("Training Data shape:", training_data_encoded.shape)
print("Testing Data shape:", testing_data_encoded.shape)
print("Saved: training-data-encoded.csv and testing-data-encoded.csv")
````

#### Step 6.2: Splitting Input Vectors and Outputs / Labels of Training Data

````python
# Separate feature vectors and target labels from training set
input_vector_train = training_data_encoded.iloc[:, :-1]
output_label_train = training_data_encoded.iloc[:, -1]

print("Input vectors (train) shape:", input_vector_train.shape)
print("Output labels (train) shape:", output_label_train.shape)
````

#### Step 6.3: Train the Support Vector Classifier

````python
"""
Purpose of this section:
-------------------------
Train the Support Vector Classifier (SVC) with gamma='auto' on the training set.
"""

print("\nTraining the Support Vector Classifier on Training Data")
print("========================================================\n")

svc_model = svm.SVC(gamma='auto', random_state=0)
svc_model.fit(input_vector_train, np.ravel(output_label_train))

print(svc_model)
````

#### Step 6.4: Save the Trained Model

````python
# Serialize and save trained model to disk
pickle.dump(svc_model, open('svc_trained_model.pkl', 'wb'))
print("Saved: svc_trained_model.pkl")
````

#### Step 6.5: Classifier Benchmark & Model Comparison (NEW)

````python
"""
Purpose of this section:
-------------------------
Train candidate models (Logistic Regression and Decision Tree) alongside SVC
on identical training data and benchmark their performance on the held-out test data.
"""

input_vector_test = testing_data_encoded.iloc[:, :-1]
output_label_test = testing_data_encoded.iloc[:, -1]

candidate_models = {
    "SVC (gamma=auto)": svc_model,
    "Logistic Regression": LogisticRegression(random_state=0, max_iter=1000).fit(input_vector_train, np.ravel(output_label_train)),
    "Decision Tree": DecisionTreeClassifier(random_state=0, max_depth=3).fit(input_vector_train, np.ravel(output_label_train)),
}

comparison_rows = []
for name, mdl in candidate_models.items():
    preds = mdl.predict(input_vector_test)
    comparison_rows.append({
        "Model": name,
        "Accuracy": round(accuracy_score(output_label_test, preds), 2),
        "Precision": round(classification_report(output_label_test, preds, output_dict=True, zero_division=0)["weighted avg"]["precision"], 2),
        "Recall": round(classification_report(output_label_test, preds, output_dict=True, zero_division=0)["weighted avg"]["recall"], 2),
        "F1-score": round(classification_report(output_label_test, preds, output_dict=True, zero_division=0)["weighted avg"]["f1-score"], 2),
    })

comparison_df = pd.DataFrame(comparison_rows)

table = PrettyTable()
table.field_names = list(comparison_df.columns)
for _, row in comparison_df.iterrows():
    table.add_row(list(row))

print("\nModel Comparison on Held-Out Testing Data:")
print(table)

plt.figure(figsize=(7, 4))
plt.bar(comparison_df["Model"], comparison_df["Accuracy"], color=['#2b5c8f', '#4682b4', '#5f9ea0'])
plt.ylim(0, 1.0)
plt.ylabel("Accuracy Score")
plt.title("Model Comparison - Accuracy on Testing Data")
for i, v in enumerate(comparison_df["Accuracy"]):
    plt.text(i, v + 0.02, str(v), ha='center', fontweight='bold')
plt.tight_layout()
plt.show()
````

### Step 7: Execute the Testing Phase

#### Step 7.1: Splitting Input Vectors and Outputs/Labels of Testing Data

````python
input_vector_test = testing_data_encoded.iloc[:, :-1]
output_label_test = testing_data_encoded.iloc[:, -1]

print("Input vectors (test) shape:", input_vector_test.shape)
print("Output labels (test) shape:", output_label_test.shape)
````

#### Step 7.2: Load the Saved Model

````python
# Load model from disk to test deployment readiness
model = pickle.load(open('svc_trained_model.pkl', 'rb'))
print("Model loaded from svc_trained_model.pkl")
````

#### Step 7.3: Evaluate the Machine Learning Model

##### Step 7.3.1: Make Predictions with the Trained Models on Testing Data

````python
# Make predictions on unseen test data
model_predictions = model.predict(input_vector_test)

pd.options.mode.chained_assignment = None
testing_data_encoded["Predictions"] = model_predictions
testing_data_encoded.to_csv("model-predictions.csv", index=False, header=True)

print("\nPredictions Returned by svc_trained_model (First 15 Test Instances):")
print("===================================================================\n")
print(testing_data_encoded.head(15))
````

#### Step 7.4: Calculate the Accuracy Score

````python
model_accuracy_score = accuracy_score(
    testing_data_encoded["HeartDisease"],
    testing_data_encoded["Predictions"]
)

print("\nAccuracy Score:")
print("===============\n")
print(round(model_accuracy_score, 2))
````

#### Step 7.5: Detailed Evaluation Metrics (Precision, Recall, F1, Confusion Matrix) (NEW)

````python
"""
Purpose of this section:
-------------------------
Calculate precision, recall, F1-score, and render a confusion matrix heatmap
to evaluate balanced clinical diagnostic efficacy.
"""

print("\nClassification Report:")
print("=======================\n")
print(classification_report(
    testing_data_encoded["HeartDisease"],
    testing_data_encoded["Predictions"],
    target_names=["No Heart Disease (0)", "Heart Disease Present (1)"]
))

conf_matrix = confusion_matrix(testing_data_encoded["HeartDisease"], testing_data_encoded["Predictions"])
conf_matrix_df = pd.DataFrame(
    conf_matrix,
    index=["Actual: No Disease", "Actual: Disease Present"],
    columns=["Predicted: No Disease", "Predicted: Disease Present"]
)

print("\nConfusion Matrix:")
print("==================\n")
print(conf_matrix_df)

plt.figure(figsize=(5, 4))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues',
            xticklabels=['No Disease (0)', 'Disease Present (1)'],
            yticklabels=['No Disease (0)', 'Disease Present (1)'])
plt.xlabel('Predicted Diagnosis')
plt.ylabel('Actual Clinical Diagnosis')
plt.title('Confusion Matrix - SVC Heart Disease Testing Data')
plt.tight_layout()
plt.show()
````
